## 7.5 — Capitalize the person

```python
things[1] = things[1].capitalize()
print(things)
```

## 9.2 — Generator function
Create `get_odds()` to yield the odd numbers from `range(10)`.
Use a `for` loop to print the third odd number produced.

**Answer:** The third odd number is **5**.

```python
def get_odds():
    for number in range(10):
        if number % 2 == 1:
            yield number

for position, odd_number in enumerate(get_odds(), start=1):
    if position == 3:
        print("Third odd number:", odd_number)
        break
```
